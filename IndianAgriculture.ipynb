```python

import os

# Load dataset
df = pd.read_csv("ICRISAT-District Level Data.csv")

# Set visualization style
sns.set_theme(style="whitegrid")

# Create folder to save charts
os.makedirs("Agriculture_Charts", exist_ok=True)

# Convert agricultural measurements to numeric
for col in df.columns:
    if col not in ["Dist Code", "Year", "State Code"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Treat -1 as missing in agricultural measurements
metric_cols = [
    c for c in df.columns
    if any(x in c.upper() for x in ["AREA", "PRODUCTION", "YIELD"])
]

df[metric_cols] = df[metric_cols].replace(-1, np.nan)

# Helper function to find columns
def get_col(crop, metric):
    for c in df.columns:
        name = c.upper()
        if crop.upper() in name and metric.upper() in name:
            return c
    raise ValueError(f"Column not found: {crop} {metric}")

# Helper function to save charts
def save_chart(number, title):
    plt.title(title, fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.savefig(
        f"Agriculture_Charts/Chart_{number}.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()
```

Chart 1 – Rice Area Distribution (Histogram)

```python
plt.figure(figsize=(10, 5))
sns.histplot(df[get_col("RICE", "AREA")].dropna(),
             bins=30, kde=True)
plt.xlabel("Rice Area (1000 ha)")
plt.ylabel("Frequency")
save_chart(1, "Distribution of Rice Cultivation Area")
```

Chart 2 – Rice Production Distribution

```python
plt.figure(figsize=(10, 5))
sns.histplot(df[get_col("RICE", "PRODUCTION")].dropna(),
             bins=30, kde=True)
plt.xlabel("Rice Production (1000 tons)")
plt.ylabel("Frequency")
save_chart(2, "Distribution of Rice Production")
```

Chart 3 – Rice Yield Distribution

```python
plt.figure(figsize=(10, 5))
sns.histplot(df[get_col("RICE", "YIELD")].dropna(),
             bins=30, kde=True)
plt.xlabel("Rice Yield (Kg per ha)")
plt.ylabel("Frequency")
save_chart(3, "Distribution of Rice Yield")
```

Chart 4 – Rice, Wheat and Cotton Area (Box Plot)

```python
area_cols = [
    get_col("RICE", "AREA"),
    get_col("WHEAT", "AREA"),
    get_col("COTTON", "AREA")
]

area_data = df[area_cols].melt(
    var_name="Crop",
    value_name="Area"
)

plt.figure(figsize=(10, 6))
sns.boxplot(data=area_data, x="Crop", y="Area")
plt.xticks(rotation=15)
plt.ylabel("Area (1000 ha)")
save_chart(4, "Comparison of Crop Cultivation Area")
```

Chart 5 – Rice Production Over the Years

```python
rice_prod = get_col("RICE", "PRODUCTION")

yearly = df.groupby("Year")[rice_prod].sum().reset_index()

plt.figure(figsize=(12, 5))
sns.lineplot(data=yearly, x="Year", y=rice_prod)
plt.xlabel("Year")
plt.ylabel("Rice Production (1000 tons)")
save_chart(5, "Year-wise Rice Production Trend")
```
